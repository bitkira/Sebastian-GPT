```python
import os
import urllib.request as request

if not os.path.exists("the-verdict.txt"):
    url = "https://raw.githubusercontent.com/rasbt/LLMs-from-scratch/refs/heads/main/ch02/01_main-chapter-code/the-verdict.txt"
    os.makedirs("./text_data")
    file_path = "./text_data/the-verdict.txt"
    request.urlretrieve(url, file_path)
```

```python
with open(file_path, "r", encoding="utf-8") as f:
    raw_text = f.read()

print(raw_text)
```

```python
import re

text = raw_text

pattern = r"([a-zA-Z']+(?:[-][a-zA-Z]+)*)|([,.!?;:\"'])|(--)"
result = re.findall(pattern, text)

# Clean up the result list by removing empty strings from the tuples
cleaned_result = [item for tuple_item in result for item in tuple_item if item]

len(cleaned_result
```

```python
cleaned_result = set(cleaned_result)
```

```python
cleaned_result = sorted(cleaned_result)
```

```python
vocab = {token:idx for idx, token in enumerate(cleaned_result)}
```

```python
def vocab2tokenid(text:str,vocab:dict):
    pattern = r"([a-zA-Z']+(?:[-][a-zA-Z]+)*)|([,.!?;:\"'])|(--)"
    result = re.findall(pattern, text)
    # Clean up the result list by removing empty strings from the tuples
    cleaned_result = [item for tuple_item in result for item in tuple_item if item]
    return [vocab[word] for word in cleaned_result]
```

```python
text = "If I could have painted that face"
tokenid = vocab2tokenid(text, vocab)
```

```python
def tokenid2vocab(tokenid:list, vocab:dict):
    text = []
    vocab = {tokenid: token for token, tokenid in vocab.items()}
    for i in tokenid:
        text.append(vocab[i])
    return "".join(text)
```

```python
tokenid2vocab(tokenid, vocab)
```
